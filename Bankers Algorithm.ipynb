```python
# Function to check system safety
def is_safe(processes, avail, max_need, allocation):
    n = len(processes)  # Number of processes
    m = len(avail)      # Number of resources
    finish = [False] * n
    safe_sequence = []
    work = avail[:]

    while len(safe_sequence) < n:
        found = False
        for i in range(n):
            if not finish[i]:
                need = [max_need[i][j] - allocation[i][j] for j in range(m)]
                if all(need[j] <= work[j] for j in range(m)):
                    for j in range(m):
                        work[j] += allocation[i][j]
                    finish[i] = True
                    safe_sequence.append(processes[i])
                    found = True
        if not found:
            return False, []

    return True, safe_sequence

# Banker's Algorithm
def bankers_algorithm(processes, avail, max_need, allocation):
    safe, safe_sequence = is_safe(processes, avail, max_need, allocation)
    if safe:
        print("The system is in a safe state.")
        print("Safe Sequence:", " -> ".join(safe_sequence))
    else:
        print("The system is not in a safe state.")

# Example usage
if __name__ == "__main__":
    processes = ["P0", "P1", "P2", "P3", "P4"]
    available = [3, 3, 2]  # Available resources
    max_need = [
        [7, 5, 3],  # Max needs for P0
        [3, 2, 2],  # Max needs for P1
        [9, 0, 2],  # Max needs for P2
        [2, 2, 2],  # Max needs for P3
        [4, 3, 3]   # Max needs for P4
    ]
    allocation = [
        [0, 1, 0],  # Allocation for P0
        [2, 0, 0],  # Allocation for P1
        [3, 0, 2],  # Allocation for P2
        [2, 1, 1],  # Allocation for P3
        [0, 0, 2]   # Allocation for P4
    ]

    bankers_algorithm(processes, available, max_need, allocation)
```
