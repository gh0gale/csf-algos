```python
class FullyAssociativeCache:
    def __init__(self, cache_size):
        self.cache_size = cache_size
        self.cache = {}  # Dictionary to store tag:data pairs
        self.access_order = []  # List to maintain access order for LRU

    def access_memory(self, memory_address, data=None):
        tag = memory_address  # In fully associative mapping, tag is the memory address itself
        if tag in self.cache:
            print(f"Cache Hit: Memory Address {memory_address} found in cache.")
            # Update access order (move tag to the end)
            self.access_order.remove(tag)
            self.access_order.append(tag)
        else:
            print(f"Cache Miss: Memory Address {memory_address} not found in cache.")
            if len(self.cache) >= self.cache_size:
                # Cache is full, apply LRU replacement
                lru_tag = self.access_order.pop(0)  # Remove the least recently used tag
                print(f"Cache is full. Replacing Memory Address {lru_tag}.")
                del self.cache[lru_tag]  # Remove the LRU block from cache
            
            # Add new block to cache
            self.cache[tag] = data
            self.access_order.append(tag)
            print(f"Memory Address {memory_address} loaded into cache.")

    def display_cache(self):
        print("\nCurrent Cache Contents:")
        for tag, data in self.cache.items():
            print(f"Tag: {tag}, Data: {data}")
        print("Access Order (most recent last):", self.access_order)


# Example usage
cache = FullyAssociativeCache(cache_size=3)  # Create a cache with 3 blocks

# Simulate memory accesses
cache.access_memory(100, "Data1")
cache.access_memory(200, "Data2")
cache.access_memory(300, "Data3")
cache.display_cache()

cache.access_memory(100)  # Cache hit
cache.access_memory(400, "Data4")  # Cache miss, triggers replacement
cache.display_cache()

cache.access_memory(300)  # Cache hit
cache.access_memory(500, "Data5")  # Cache miss, triggers replacement
cache.display_cache()
```
