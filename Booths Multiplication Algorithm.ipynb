```python
def decimal_to_binary(n, size):
    
    if n < 0:
        n = (1 << size) + n
    return format(n, '0{}b'.format(size))

def binary_to_decimal(b):
    
    if b[0] == '1':  
        return int(b, 2) - (1 << len(b))
    else:
        return int(b, 2)

def arithmetic_right_shift(a, q, qn):
    
    combined = a + q + qn
    shifted = combined[0] + combined[:-1]
    return shifted[:len(a)], shifted[len(a):len(a) + len(q)], shifted[-1]

def booth_multiplication(multiplicand, multiplier, register_size):
    
    
    a = '0' * register_size
    q = decimal_to_binary(multiplier, register_size)
    m = decimal_to_binary(multiplicand, register_size)
    qn = '0'
    
   
    neg_m = decimal_to_binary(-multiplicand, register_size)
    
    print(f"Initial values:\nA: {a}\nQ: {q}\nM: {m}\n-M: {neg_m}\nQn: {qn}\n")
    
    for i in range(register_size):
        print(f"Step {i+1}:")
        
        if q[-1] == '1' and qn == '0':  # Q0 = 1 and Q-1 = 0, A = A - M
            a = decimal_to_binary(binary_to_decimal(a) + binary_to_decimal(neg_m), register_size)
            print(f"  A = A - M: {a}")
        
        elif q[-1] == '0' and qn == '1':  # Q0 = 0 and Q-1 = 1, A = A + M
            a = decimal_to_binary(binary_to_decimal(a) + binary_to_decimal(m), register_size)
            print(f"  A = A + M: {a}")
        
        
        a, q, qn = arithmetic_right_shift(a, q, qn)
        print(f"  After shift: A = {a}, Q = {q}, Qn = {qn}\n")
    
   
    result = a + q
    return binary_to_decimal(result)


multiplicand = int(input("Enter the multiplicand (in decimal): "))
multiplier = int(input("Enter the multiplier (in decimal): "))
register_size = int(input("Enter the register size (in bits): "))


result = booth_multiplication(multiplicand, multiplier, register_size)
print(f"\nResult of Booth's Multiplication: {result}")
```
